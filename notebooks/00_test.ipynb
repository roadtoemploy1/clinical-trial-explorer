```python
import json
import requests
url = "https://clinicaltrials.gov/api/v2/studies"
params = {"query.cond": "non-small cell lung cancer", "pageSize": 5}
response = requests.get(url, params=params, timeout=30)
response.raise_for_status()
data = response.json()
```

```python
for study in data["studies"]:
    ps = study["protocolSection"]
    ident = ps["identificationModule"]
    stat = ps["statusModule"]
    sponsor = ps["sponsorCollaboratorsModule"]["leadSponsor"]
    design = ps["designModule"]
    phases = ", ".join(design.get("phases",[])) or "None"
    
    print(f"{ident['nctId']}  {ident['briefTitle']}")
    print(f"  Type:    {design.get('studyType')}")
    print(f"  Phases:  {phases}")
    print(f"  Status:  {stat.get('overallStatus')}")
    print(f"  Sponsor: {sponsor.get('name')} ({sponsor.get('class')})")
    print()
```
